## 📂 Bước 2: Upload file `main_flet.py`

```python
from google.colab import files
import os
if os.path.exists('main_flet.py'):
    os.remove('main_flet.py')
print('👉 Chọn file main_flet.py\n')
uploaded = files.upload()
if os.path.exists('main_flet.py'):
    print(f'\n✅ Đã upload: {os.path.getsize("main_flet.py")/1024:.1f} KB')
else:
    print('\n❌ Upload lại nhé.')
```

## 🔨 Bước 3: Build APK

⏱️ **Lần đầu ~10-20 phút** (Flet tự động tải Android SDK + NDK).

```python
import time
t0 = time.time()
!yes | flet build apk 2>&1 | tee build.log
elapsed = time.time() - t0
print(f'\n⏱️ Thời gian build: {int(elapsed//60)} phút {int(elapsed%60)} giây')
```

## 📥 Bước 4: Tải APK về

```python
import os, glob
from google.colab import files
apks = glob.glob('build/apk/*.apk') + glob.glob('dist/*.apk')
print('Tìm thấy các APK:')
for apk in apks:
    print(f'  📦 {apk} ({os.path.getsize(apk)/1024/1024:.2f} MB)')
if apks:
    target = min(apks, key=os.path.getsize)
    print(f'\n🎯 Tải: {target}')
    files.download(target)
    print('\n✅ Đã tải APK thành công!')
    print('👉 Copy file .apk vào điện thoại Android → mở file → cài đặt')
else:
    print('\n❌ Không tìm thấy APK. Kiểm tra log bằng: !tail -50 build.log')
```
